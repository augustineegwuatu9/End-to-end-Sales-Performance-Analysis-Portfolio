```python
## Clean the columns
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
## Columns cleaned
df.columns 
```

```python
#Fix date column to YYYY-MM-DD format
df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
## Checked to see that it worked
df.head() 
```

```python
## Create time features
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name
```

```python
## Create revenue column
df['revenue'] = df['quantity'] * df['unit_price']
```

```python
## Check for missing values
df.duplicated().sum()
```

```python
## code to rune if duplicate exists but none here
df = df.drop_duplicates()
```

```python
## Revenue by category
df.groupby('product_category')['revenue'].sum().sort_values(ascending=False)
```

```python
## Revenue by states
df.groupby('region')['revenue'].sum().sort_values(ascending=False)
```

```python
## Monthly trend
df.groupby(['year', 'month'])['revenue'].sum() 
```

```python
## Export the cleaned data for SQL and Tableau
df.to_csv("cleaned_sales_data.csv", index=False)
```

```python
## 2. Advanced Analysis Begins
## Customer value analysis to identify high-value customers and measuring customer lifetime value
customer_value = (
    df.groupby('customer_id')
    .agg(
        total_revenue=('revenue', 'sum'),
        total_orders=('order_id', 'nunique')
    )
    .sort_values(by='total_revenue', ascending=False)
)

customer_value.head()
```

```python
## Create customer segment
customer_value['segment'] = pd.qcut(
    customer_value['total_revenue'],
    q=4,
    labels=['Low', 'Mid', 'High', 'VIP']
)
```

```python
## Revenue by segment
customer_value.groupby('segment')['total_revenue'].sum()
```

```python
## Merge back to main daaset
df =  df.merge(customer_value[['segment']],
on='customer_id', how='left')


## Dataset now has trasaction data and customer segment
```

```python
## Ivnvestigate product performance to discover high revenue associated with low orders and high orders with low revenue
## Prduct category's revenue and num_orders
product_perf = (
    df.groupby('product_category')
    .agg(
        revenue=('revenue', 'sum'),
        orders=('order_id', 'count')
    )
    .sort_values(by='revenue', ascending=False)
)
product_perf
```

```python
## Regional performance to find out underperforming states
region_perf = (
    df.groupby('region')['revenue']
    .sum()
    .sort_values(ascending=False)
)
region_perf
```

```python
## Time analysis - Used for line chart in Tableau
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month

monthly_trend = (
    df.groupby(['year', 'month']) ['revenue']
    .sum()
    .reset_index()
)
```

```python
## Export finalized sale dataset
df.to_csv('final_sales_dataset.csv', index=False)
```
